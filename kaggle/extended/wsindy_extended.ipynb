```python
import os, sys, subprocess, base64, io, zipfile
from pathlib import Path
for name in ('OPENBLAS_NUM_THREADS', 'OMP_NUM_THREADS', 'MKL_NUM_THREADS'):
    os.environ[name] = '1'
root = Path('/kaggle/working/wsindy_research')
root.mkdir(parents=True, exist_ok=True)
payload = ''
with zipfile.ZipFile(io.BytesIO(base64.b64decode(payload))) as archive:
    for entry in archive.infolist():
        destination = (root / entry.filename).resolve()
        if not destination.is_relative_to(root.resolve()):
            raise ValueError('Archive path outside workspace')
    archive.extractall(root)
print('Source unpacked:', root)
```

```python
import importlib.metadata
print({p: importlib.metadata.version(p) for p in ('numpy', 'scipy', 'matplotlib', 'pytest')})
subprocess.run([sys.executable, '-m', 'pytest', '-q', 'tests'], cwd=root, check=True)
```
